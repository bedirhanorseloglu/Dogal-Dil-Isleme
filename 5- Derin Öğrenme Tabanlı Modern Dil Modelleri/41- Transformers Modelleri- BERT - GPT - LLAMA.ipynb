# 2- Transformers Nedir?

- RNN -> LSTM -> Transformer (Her çıkan bir öncekinin eksiğini gidermeye yönelik)
- LSTM'in de eksik yönleri bağlamı değerlendirememe ve sıralı işlem yapmasıydı.



### 2.1 Temel Yapı

Transformer:
- Encoder + Decoder mimarisi
- Paralel işlem yapabilir
- Büyük veri üzerinde verimlidir

#### Ana bileşenler:
- Encoder (Kodlayıcı)
- Decoder (Kod Çözücü)

---

### 2.2 Encoder Bileşenleri

Encoder giriş dizisini işler ve temsil üretir.

#### Her encoder katmanı:

##### 🔹 2.2.1 Self-Attention

- Her kelime diğer kelimelerle ilişki kurar
- Q (Query), K (Key), V (Value) vektörleri oluşturulur

###### Süreç:
1. Girdi cümle alınır
2. Kelimeler embedding’e çevrilir
3. Q, K, V hesaplanır
4. Softmax ile ağırlıklandırma yapılır
5. Ağırlıklı vektörler toplanır

📌 Amaç:
- Önemli kelimeleri öne çıkarmak
- Önemsizleri bastırmak

---

##### 🔹 2.2.2 Feed Forward Neural Network

- Self-attention çıktısı işlenir
- 2 katmanlı fully-connected yapı
- Aktivasyon: genelde ReLU

📌 Amaç:
- Daha soyut ve güçlü temsil üretmek

---

### 2.3 Decoder Bileşenleri

Decoder:
- Self-attention
- Encoder-Decoder attention
- Feed-forward

#### Ek Katman:
- Encoder çıktısına odaklanır

# 3- BERT

### 3.1 Genel Bakış

- Google tarafından 2018’de geliştirildi
- Sadece **Encoder** kullanır
- Çift yönlü çalışır (bidirectional) (metni hem sağdan sola hem de soldan sağa okuyarak bağlamı anlamaya çalışır.)

📌 GPT vs BERT:
- GPT → tek yönlü
- BERT → çift yönlü

---

### 3.2 Model Boyutları

| Model        | Katman | Parametre |
|-------------|--------|----------|
| BERT Base   | 12     | ~110M    |
| BERT Large  | 24     | ~340M    |

---

### 3.3 Girdi Temsilleri

BERT girişte 3 embedding kullanır:

- **Token Embedding**
- **Segment Embedding**
- **Position Embedding**

---

### 3.4 Eğitim Görevleri

#### 🔹 Masked Language Model (MLM)

- Kelimelerin %15’i maskelenir
- Model tahmin eder

#### 🔹 Next Sentence Prediction (NSP)

- İki cümle arasındaki ilişki öğrenilir
- Binary classification

---

### 3.5 Kullanım Alanları

- Metin sınıflandırma
- Duygu analizi
- Soru cevaplama
- NER (Named Entity Recognition)
- Cümle benzerliği


## 4- GPT




### 4.1 Genel Bakış

- GPT, metin üretme ve dil modelleme için kullanılan bir dil modelidir.
- Metni sadece soldan sağa doğru okur ve bir kelimeyi tahmin etmek için önceki kelimelere dayanır.
- Metin üretme, öykü yazma, yaratıcı içerik üretme gibi alanalarda kullanılır



---

### 4.2 GPT Mimarisi

- Sadece **decoder** kullanır. Bu yapı dil üretiminde ilei doğru yani sola doğru işlem yapar yani soldaki kelimeleri alarak ileri (sağa) doğru işlem yapar. Bu yüzden bir kelimeyi tahmin ederken sadece o kelimeden önceki kelimeleri dikkate alır.



- **Otokorelasyonlu (autoagressivve) bir yaklaşım** sergiler. Yani bir kelimeyi tahmin ederken daha önce ürettiği kelimeleri dikkate alır.
- BERT'in aksine tek aşamalı bir eğitimden geçer -> NSP

## 5- LLAMA

### 5.1 Genel Bakış

- LLaMA, Meta tarafından geliştirlen büyük dil modeli ailesidir.
- LLaMA, büyük veri kümeleri üzerinde eğitilmiş olup, GPT-3 gibi modellerler kıyaslandığında daha az parametre ile benzer performansı sağlamayı hedefler.



- Amacı, akademik araştırmacılara büyük dil modellerinin eğitimini daha erişilebilir hale getirmek ve daha verimli modeller sunmaktır.

---

### 5.2 LLAMA Mimarisi

- Transformer tabanlıdır. Transformer mimarisini kullanır ancak diğer modellere kıyasla daha verimli bir eğitim sürecine sahiptir.
- Diğer modellere göre daha hafif ve verimlidir. Daha az parametre kullanarak GPT-3 gibi modellerle benzer performans sergilemek üzere tasarlanmıştır. 

| Model     | Parametre |
|-----------|----------|
| GPT-3     | ~175 Milyar|
| LLaMA     | ~13 Milyar |



- Buradaki artık bağlantılar (kırmızı ok) transformer'da da olduğu gibi bilginin iletimini yani hafıza sağlar ve kaybolan gradyan probleminin oluşmasını engellemeyi amaçlar.
